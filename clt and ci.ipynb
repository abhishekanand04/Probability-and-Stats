```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
!wget --no-check-certificate https://drive.google.com/uc?id=1Mrt008vkE4nVb1zE4f06_rtq70QPfkIo -O weight-height.csv
```

```python
df_hw = pd.read_csv('weight-height.csv')
df_hw.head()
```

```python
df = df_hw['Height']
df
```

```python
df.mean()
```

```python
df.std()
```

```python
sns.histplot(df,kde=True)
```

```python
#clt here
#sample size=5
df.sample(5)
```

```python
np.mean(df.sample(5))
```

```python
#repeat this 10k times => sample of size 5 => sample means
sample_5 = [np.mean(df.sample(5)) for i in range(10000)]
```

```python
sample_5
```

```python
sns.histplot(df, kde = True)
```

```python
sns.histplot(sample_5, kde = True)
```

```python
np.mean(sample_5)
```

```python
np.std(sample_5)
```

```python
#repeat this 10k times => sample of size 20 => sample means
sample_20 = [np.mean(df.sample(20)) for i in range(10000)]
```

```python
sns.histplot(sample_20,kde=True)
```

```python
np.mean(sample_20)
```

```python
np.std(sample_20)
```

```python
#comparison in a single cell
# population mean
mu = df.mean()

# population SD
sigma = df.std()

# mean of sample distributions having sample size = 5
mu_5 = np.mean(sample_5)

# SD of sample distributions having sample size = 5
sigma_5 = np.std(sample_5)

# mean of sample distributions having sample size = 20
mu_20 = np.mean(sample_20)

# SD of sample distributions having sample size = 20
sigma_20 = np.std(sample_20)
```

```python
mu,mu_5,mu_20
```

```python
sigma,sigma_5,sigma_20
```

```python
#given sigma, sample size =n
sigma/np.sqrt(5)
```

```python
sigma/np.sqrt(20)
```
