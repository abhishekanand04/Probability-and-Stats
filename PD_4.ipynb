```python
#Bootstrapping
survey_1 = [35,36,33,37,34,35]
survey_2 = [20,37,17,50,53,33]
#pop mean => no, pop std-dev => no
#sample size => small
```

```python
np.mean(survey_1)
```

```python
np.mean(survey_2)
```

```python
sample1 = np.random.choice(survey_1,size=6)
sample1
```

```python
np.mean(sample1)
```

```python
boostrapped_mean_survey_1 = []
for i in range(10000):
  boostrapped_sample = np.random.choice(survey_1,size=6)
  bootsrapped_mean = np.mean(boostrapped_sample)
  boostrapped_mean_survey_1.append(bootsrapped_mean)
```

```python
boostrapped_mean_survey_1
```

```python
sns.histplot(boostrapped_mean_survey_1,kde=True)
```

```python
boostrapped_mean_survey_2 = []
for i in range(10000):
  boostrapped_sample = np.random.choice(survey_2,size=6)
  bootsrapped_mean = np.mean(boostrapped_sample)
  boostrapped_mean_survey_2.append(bootsrapped_mean)
```

```python
sns.histplot(boostrapped_mean_survey_2,kde=True)
```

```python
#CI
x1 = np.percentile(boostrapped_mean_survey_1,2.5)
x2 = np.percentile(boostrapped_mean_survey_1,97.5)
x1,x2
```

```python
x1 = np.percentile(boostrapped_mean_survey_2,2.5)
x2 = np.percentile(boostrapped_mean_survey_2,97.5)
x1,x2
```
